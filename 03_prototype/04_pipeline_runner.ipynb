```python
import os
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import time

def run_notebook(notebook_path):
    print(f"[{time.strftime('%H:%M:%S')}] Starting execution of: {notebook_path}")
    start_time = time.time()
    
    with open(notebook_path, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
        
    ep = ExecutePreprocessor(timeout=None, kernel_name='python3')
    
    try:
        ep.preprocess(nb, {'metadata': {'path': './'}})
        
        with open(notebook_path, 'w', encoding='utf-8') as f:
            nbformat.write(nb, f)
            
        elapsed_time = time.time() - start_time
        print(f"[{time.strftime('%H:%M:%S')}] Successfully finished {notebook_path} in {elapsed_time:.2f} seconds.\n")
        return True
        
    except Exception as e:
        print(f"\n Error executing {notebook_path}!")
        print(f"Exception: {str(e)}")
        return False

notebooks_to_run = [
    "01_eda_and_preparation.ipynb",
    "02_rasterization_pipeline.ipynb",
    "03_prototype.ipynb"
]

print("Starting End-to-End Prototype Pipeline...\n" + "-"*50)

# Run them one by one
for nb_file in notebooks_to_run:
    if not os.path.exists(nb_file):
        print(f"Error: Could not find {nb_file}. Make sure you are running this from the 'prototype' folder.")
        break
        
    success = run_notebook(nb_file)
    
    if not success:
        print("Pipeline aborted due to error.")
        break

if success:
    print("-"*50 + "\nEntire Pipeline Completed Successfully!")
```
